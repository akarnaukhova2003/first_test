```python
from Bio import SeqIO

with open("AV_2.gb") as handle:
    records = list(SeqIO.parse(handle, "gb"))
    for rec in records:
        accession = rec.annotations["accessions"][0] 
        for feature in rec.features:
            if feature.type == "CDS":
                print(accession, feature.location)
```

```python
from Bio import SeqIO

input_file = "AV_2.gb"
output_file = "AV_2_revcomp.gb"
negatives_file = "negative_ids.txt"

records = []
negative_info = []

with open(input_file) as handle:
    for rec in SeqIO.parse(handle, "gb"):
        accession = rec.annotations["accessions"][0]
        has_negative_cds = False
        coords = []
        for feature in rec.features:
            if feature.type == "CDS" and feature.location.strand == -1:
                has_negative_cds = True
                coords.append(str(feature.location))

        if has_negative_cds:
            print(f"{accession} → (-) цепь → делаем reverse complement")
            rec.seq = rec.seq.reverse_complement()
            negative_info.append((accession, coords))
        else:
            print(f"{accession} → (+) цепь → без изменений")
        records.append(rec)

with open(output_file, "w") as out_handle:
    SeqIO.write(records, out_handle, "gb")
with open(negatives_file, "w") as f:
    for acc, coords in negative_info:
        f.write(acc + "\t" + ", ".join(coords) + "\n")
```
