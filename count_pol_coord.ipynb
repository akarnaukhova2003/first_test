```python
import csv
import matplotlib.pyplot as plt

lengths = [] 
lengths_less_1000 = []
with open("AV_2_modified_orf.txt") as file:
    reader = csv.DictReader(file, delimiter=',') 
    for row in reader:
        coord = row['1B']
        if coord != 'NA-NA' and '-' in coord:
            try:
                start, end = map(int, coord.split('-'))
            except ValueError:
                continue
            length = end - start + 1
            if length < 1000 and length > 600 :
                print(row['id'])
            if length < 1000:
                lengths_less_1000.append(length)
            lengths.append(length)
print(f"Количество генов 1B: {len(lengths)}")
thresholds = [500, 1400, 1500]
for t in thresholds:
    print(f"Количество генов длиной > {t} nt: {sum(l > t for l in lengths)}")

plt.figure(figsize=(10,6))
plt.hist(lengths, bins=100, color='skyblue', edgecolor='black')
plt.xlabel("Длина ORF 1B (nt)")
plt.ylabel("Количество генов")
plt.title("Распределение длин ORF 1B")
plt.grid(axis='y', alpha=0.75)
plt.show()
```
