```python
import csv

file1 = "all_cds_names_2.txt"
file2 = "mav_orf.csv"
outfile = "1ab_2.txt"

mapping = {}
with open(file2, encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        if len(row) == 2:
            key, value = row
            mapping[key.strip()] = value.strip()

with open(file1, encoding="utf-8") as f_in, open(outfile, "w", encoding="utf-8") as f_out:
    for line in f_in:
        word = line.strip()
        if word in mapping:
            f_out.write(f"{word},{mapping[word]}\n")
        else:
            f_out.write(word + "\n")
```

```python
import csv

file1 = "1ab_2.txt"
file2 = "1b.csv"
outfile = "1ab_2_final.txt"

# читаем словарь из второго файла
mapping = {}
with open(file2, encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        if len(row) == 2:
            key, value = row
            mapping[key.strip()] = value.strip()

# обрабатываем первый файл
with open(file1, encoding="utf-8") as f_in, open(outfile, "w", encoding="utf-8") as f_out:
    for line in f_in:
        word = line.strip()
        # если в строке уже есть запятая → оставляем
        if "," in word:
            f_out.write(word + "\n")
        else:
            # если слово есть в словаре → добавляем через запятую
            if word in mapping:
                f_out.write(f"{word},{mapping[word]}\n")
            else:
                f_out.write(word + "\n")
```

```python
import csv

file_in = "1ab_2_final.txt"
file_out = "1ab_2_final.csv"
with open(file_in, encoding="utf-8") as f_in, open(file_out, "w", newline="", encoding="utf-8") as f_out:
    writer = csv.writer(f_out, delimiter=",")
    for line in f_in:
        # убираем пробелы по краям
        line = line.strip()
        if not line:
            continue  # пропускаем пустые строки

        # если в строке уже есть запятая → используем как есть
        if "," in line:
            row = [x.strip() for x in line.split(",")]
        else:
            # если запятой нет → пишем как одна колонка
            row = [line]

        writer.writerow(row)
```
